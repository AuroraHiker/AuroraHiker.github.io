```python
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# 加载数据
iris = load_iris()
X = iris.data          # 特征矩阵 (150, 4)
y = iris.target        # 目标标签 (150,)
feature_names = iris.feature_names
target_names = iris.target_names

# 设置绘图风格
plt.rcParams.update({
    'font.family': 'Times New Roman',   # 字体设置
    'font.weight': 'bold',
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold',
    "font.size": 16,                 # 基础字号
    "axes.labelsize": 16,            # 轴标签字号
    "xtick.labelsize": 0,           # X轴刻度字号
    "ytick.labelsize": 0,           # Y轴刻度字号
    "legend.fontsize": 14,           # 图例字号
    # 刻度线设置
    'xtick.minor.size': 0,              # 关闭次要刻度
    'ytick.minor.size': 0,
    'xtick.major.size': 0,              # 主刻度长度
    'ytick.major.size': 0,
    'xtick.major.width': 1,           # 主刻度粗细
    'ytick.major.width': 1,
    'xtick.minor.width': 1,           # 次刻度粗细
    'ytick.minor.width': 1,
    # 坐标轴与边框设置
    'axes.grid': False,                 # 关闭网格线
    'axes.edgecolor': 'gray',          # 轴线颜色
    'axes.linewidth': 1,              # 轴线粗细
    'axes.spines.left': True,           # 显示所有边框
    'axes.spines.right':True,
    'axes.spines.top': True,
    'axes.spines.bottom': True,
    'mathtext.default' : 'it',          # 允许使用数学表达式
    'mathtext.fontset': 'stix',      # 设置数学公式字体为stix，类似于 Times New Roman
    # 其他设置
    'axes.unicode_minus': False         # 解决负号显示问题
})
```

```python
df = pd.DataFrame(X, columns=feature_names)
df['species'] = y

# 绘制花瓣长度 vs 花瓣宽度
plt.figure(figsize=(8, 6))
for species in range(3):
    subset = df[df['species'] == species]
    plt.scatter(subset['petal length (cm)'], subset['petal width (cm)'], 
                label=target_names[species])
plt.xlabel('Petal length (cm)')
plt.ylabel('Petal width (cm)')
plt.legend(frameon=False)
plt.title('Iris Dataset - Petal Feature Distribution') 
plt.savefig('./iris数据概况.png', dpi=300, bbox_inches='tight')
plt.show()
```
