```python
import numpy as np
import matplotlib.pyplot as plt

# 设置绘图风格
plt.rcParams.update({
    'font.family': 'Times New Roman',   # 字体设置
    'font.weight': 'bold',
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold',
    "font.size": 16,                 # 基础字号
    "axes.labelsize": 16,            # 轴标签字号
    "xtick.labelsize": 0,           # X轴刻度字号
    "ytick.labelsize": 0,           # Y轴刻度字号
    "legend.fontsize": 14,           # 图例字号
    # 刻度线设置
    'xtick.minor.size': 0,              # 关闭次要刻度
    'ytick.minor.size': 0,
    'xtick.major.size': 0,              # 主刻度长度
    'ytick.major.size': 0,
    'xtick.major.width': 1,           # 主刻度粗细
    'ytick.major.width': 1,
    'xtick.minor.width': 1,           # 次刻度粗细
    'ytick.minor.width': 1,
    # 坐标轴与边框设置
    'axes.grid': False,                 # 关闭网格线
    'axes.edgecolor': 'gray',          # 轴线颜色
    'axes.linewidth': 1,              # 轴线粗细
    'axes.spines.left': False,           # 显示所有边框
    'axes.spines.right': False,
    'axes.spines.top': False,
    'axes.spines.bottom': False,
    'mathtext.default' : 'it',          # 允许使用数学表达式
    'mathtext.fontset': 'stix',      # 设置数学公式字体为stix，类似于 Times New Roman
    # 其他设置
    'axes.unicode_minus': False         # 解决负号显示问题
})

# 生成模拟数据：复杂度从 0 到 10
x = np.linspace(0, 10, 200)

# 训练误差：随复杂度增加单调递减（模拟指数衰减）
train_err = np.exp(-x / 2) + 0.02 * np.random.randn(len(x))  # 加一点噪声让曲线更自然

# 测试误差：先降后升（模拟 U 形）
test_err = np.exp(-x / 2) + 0.08 * (x - 4) ** 2 + 0.02 * np.random.randn(len(x))

# 平滑处理（可选）
from scipy.ndimage import gaussian_filter1d
train_err = gaussian_filter1d(train_err, sigma=2)
test_err = gaussian_filter1d(test_err, sigma=2)

# 绘图
fig, ax = plt.subplots(figsize=(7.5, 4.5))

# 绘制两条曲线
ax.plot(x, train_err, color='lightblue', linewidth=3, label='Training error')
ax.plot(x, test_err, color='plum', linewidth=3, label='Test error')



# 添加垂直虚线表示最佳复杂度区域中心
ax.axvline(x=4, linestyle='--', color='gray', alpha=0.7)

# 添加文字标注
ax.text(1.2, -0.05, 'Underfitting', fontsize=12, ha='center')
ax.text(4.0, -0.05, 'Optimal', fontsize=12, ha='center')
ax.text(7.5, -0.05, 'Overfitting', fontsize=12, ha='center')

# 轴标签和标题
ax.set_xlabel('Model Complexity', fontsize=14,labelpad=20)
ax.set_ylabel('Error', fontsize=14)
ax.set_title('Overfitting vs. Underfitting', fontsize=16)

# 图例
ax.legend(loc='center right', bbox_to_anchor=(1, 0.9), fontsize=12, frameon=False)

# 标记区域（用垂直色块）
ax.axvspan(0, 2.5, alpha=0.15, color='green', label='Underfitting zone')
ax.axvspan(2.5, 5.5, alpha=0.15, color='yellow', label='Optimal zone')
ax.axvspan(5.5, 10, alpha=0.15, color='red', label='Overfitting zone')
# 设置坐标轴范围
ax.set_xlim(0, 10)
ax.set_ylim(0, 0.9)

# 添加网格（可选）
ax.grid(True, linestyle=':', alpha=0.5)

# 显示
plt.tight_layout()
# plt.show()

# 如果需要保存图片，取消下面注释
plt.savefig('./overfitting_underfitting.png', dpi=300, bbox_inches='tight')
```
